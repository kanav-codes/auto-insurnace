# ***1. Executive Summary & One-Page Overview***

Key Takeaway 1: High Retention Base with High-Risk Segments

Over 85% of our policyholders (Response = No / retained) form a stable base, but a distinct cluster shows high claim frequencies tied to specific vehicle classes and low policy tenure.

Key Takeaway 2: Primary Drivers of Customer Lifetime Value (CLV)

Monthly Premium Auto and Total Claim Amount are the strongest predictors of CLV, meaning pricing strategy directly impacts long-term customer value.

Key Takeaway 3: Churn Vulnerability in Early Tenure

Newer customers with multiple open claims or high monthly premiums exhibit a higher propensity to churn.

Recommended Action:

Implement targeted retention discounts and proactive customer service check-ins for policyholders within their first 12–24 months who hold high monthly premiums or have filed more than 2 claims.

# ***2. Python Code for Visualizations & Dashboard Structure***

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel("AutoInsuranceClaims2024.xlsx")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean aesthetic style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.family': 'sans-serif', 'font.size': 11})

# --- Chart 1: Customer Churn vs Retention Distribution ---
plt.figure(figsize=(8, 5))
churn_counts = df['Response'].value_counts()
ax = sns.barplot(x=churn_counts.index, y=churn_counts.values, palette="Blues_d")
plt.title("Distribution of Customer Churn Status", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Response / Churn Status", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)
plt.bar_label(ax.containers[0], fmt='%d', padding=3)
plt.ylim(0, max(churn_counts.values) * 1.15)
sns.despine(top=True, right=True)
plt.show()

```

```python
# --- Chart 2: Monthly Premium vs Customer Lifetime Value (CLV) ---
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='Monthly Premium Auto',
    y='Customer Lifetime Value',
    hue='Response',
    alpha=0.7,
    palette="Set1"
)
plt.title("Customer Lifetime Value vs. Monthly Premium Auto", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Monthly Premium Auto ($)", fontsize=12)
plt.ylabel("Customer Lifetime Value ($)", fontsize=12)
plt.legend(title="Churn Status", loc="upper left")
sns.despine(top=True, right=True)
plt.show()

```

```python
# --- Chart 3: Total Claim Amount Distribution by Coverage Type ---
plt.figure(figsize=(10, 5))
sns.boxplot(
    data=df,
    x='Coverage',
    y='Total Claim Amount',
    palette="Pastel1"
)
plt.title("Claim Amount Distribution Across Coverage Tiers", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Coverage Type", fontsize=12)
plt.ylabel("Total Claim Amount ($)", fontsize=12)
sns.despine(top=True, right=True)
plt.show()
```
