```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,f1_score
```

```python
n_estimators_list=[100,200,300,400,500]
max_depth_list=[5,10,15,20]
results=[]
for n_trees in n_estimators_list:
  for depth in max_depth_list:
    rf=RandomForestClassifier(n_estimators=n_trees,max_depth=depth,random_state=42,class_weight='balanced')
    rf.fit(x_train,y_train)
    y_pred=rf.predict(x_test)
    accuracy=accuracy_score(y_test,y_pred)
    recall=recall_score(y_test,y_pred)
    precision=precision_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    results.append({'Trees':n_trees,'Depth':depth,'Accuracy':accuracy,'Recall':recall,'Precision':precision,'F1 Score':f1})
result_df=pd.DataFrame(results)
result_df=result_df.sort_values(by=['Recall','Accuracy'],ascending=False)
print(result_df.head(20))
```

```python
from sklearn.cluster import KMeans
wcss=[]
for k in range(1,16):
  kmeans=KMeans(n_clusters=k,random_state=42,n_init='auto')
  kmeans.fit(scaled_data)
  wcss.append(kmeans.inertia_)
plt.figure(figsize=(8,6))
plt.plot(range(1,16),wcss,marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method For Optimal K')
plt.show()
```

## 4. CLV Regression & Evaluation

```python
import numpy as np
target_column = df_encoded['Customer Lifetime Value']
y_log = np.log1p(target_column)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df_encoded.drop(columns=['Customer Lifetime Value'])
```

```python
x_train_reg, x_test_reg, y_train_reg, y_test_reg = train_test_split(
    x, y_log, test_size=0.2, random_state=42
)
```

```python
from sklearn.ensemble import RandomForestRegressor

reg_model = RandomForestRegressor(n_estimators=300, max_depth=15, random_state=42)
reg_model.fit(x_train_reg, y_train_reg)


y_pred_log = reg_model.predict(x_test_reg)
y_pred = np.expm1(y_pred_log)
y_test_actual = np.expm1(y_test_reg)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred))
mae = mean_absolute_error(y_test_actual, y_pred)
r2 = r2_score(y_test_actual, y_pred)

print(f'RMSE: {rmse}')
print(f'MAE: {mae}')
print(f'R2 Score: {r2}')
```

# ***CLV Regression:***
Analysis: Implemented a Random Forest Regressor yielding an $R^2$ Score of 0.958, confirming that 95.8% of the variance in CLV is explained by our feature set. Feature importance indicates Number of Policies (67.6%) and Monthly Premium (21.0%) as primary growth drivers.

Uncertainty Quantification: Evaluated model error bounds yielding an RMSE of 1822.89 and an MAE of 486.91, explicitly quantifying prediction variance on the original monetary scale via inverse log-transformation (np.expm1).

# ***Confounders, Limitations, and Data Biases***
Geographic / Selection Bias: The underlying dataset leans heavily toward specific regions (e.g., observations predominantly centered around states like Oregon), meaning model weights and feature importances might exhibit localized biases and may not generalize seamlessly to a nationwide or international insurance market.

Unobserved Confounders: Economic factors such as inflation, sudden regional policy changes, or competitor pricing strategies are omitted from the static dataset, which could alter actual customer lifetime values and churn trajectories in a live production environment.
